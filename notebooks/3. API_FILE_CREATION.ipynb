# API FILE CREATION

```python
from dotenv import load_dotenv
import os
```

```python
# Delete .env file
if os.path.exists('.env'):
    os.remove('.env')
    print(" .env file deleted")
else:
    print(" .env file doesn't exist")

# Verify it's gone
if not os.path.exists('.env'):
    print(" Confirmed: .env file is gone")
```

```python
# Create .env file for Groq API

env_content = """GROQ_API_KEY= ********* """

# Write to file
with open('.env', 'w') as f:
    f.write(env_content)

print(" .env file created for Groq API!")
print(" Location:", os.path.abspath('.env'))
```

```python
# Load the .env file
load_dotenv()

# Check if it worked
api_key = os.getenv('GROQ_API_KEY')

if api_key:
    print(" SUCCESS!")
    print(f" Groq API Key loaded: {api_key[:20]}...")
else:
    print(" FAILED - API key not found")
    print("Make sure:")
    print("  1. .env file is in your project folder")
    print("  2. Format is: GROQ_API_KEY=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")
    print("  3. No spaces around the = sign")
```
