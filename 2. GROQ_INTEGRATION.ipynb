# GROQ API SETUP

```python
from groq import Groq
import os
from dotenv import load_dotenv
```

```python
# Load API key
load_dotenv()
api_key = os.getenv('GROQ_API_KEY')

# Create client
client = Groq(api_key=api_key)

# List available models
print("=" * 80)
print("AVAILABLE GROQ MODELS")
print("=" * 80)

models = client.models.list()
for model in models.data:
    print(f" {model.id}")
```

```python
print("=" * 80)
print("STEP 1: LOADING GROQ API CONFIGURATION")
print("=" * 80)

# Load API key
load_dotenv()
api_key = os.getenv('GROQ_API_KEY')

print(" API Key loaded successfully!")
print(f"   API Key: {api_key[:5]}...")

# Create Groq client
client = Groq(api_key=api_key)

print("\n" + "=" * 80)
print("STEP 2: CREATE GROQ CLIENT")
print("=" * 80)

print(" Groq client created")
print(" Ready to use Llama 3.3 model")

print("\n" + "=" * 80)
print("STEP 3: TEST CONNECTION")
print("=" * 80)

# Test the connection
test_prompt = "Say 'Groq API is working!' in exactly these words."

try:
    response = client.chat.completions.create(
        model="llama-3.3-70b-versatile",
        max_tokens=100,
        messages=[
            {"role": "user", "content": test_prompt}
        ]
    )
    print(" SUCCESS!")
    print(f"Response: {response.choices[0].message.content}\n")
    
except Exception as e:
    print(f" ERROR: {e}")

print("=" * 80)
print("GROQ API IS READY!")
print("=" * 80)
print("\nModel: llama-3.3-70b-versatile")
print("Features:")
print("   Latest Llama 3.3 model")
print("   Excellent at instruction-following")
print("   Great for text classification")
print("   100% FREE (no credit card)")
print("   Generous rate limits")
print("   Fast processing")
```
